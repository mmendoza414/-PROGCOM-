```python
agenda = [] # Inicializa la lista de la agenda

while True:
    # Muestra el menú de opciones
    print("\n--- Menú de la Agenda ---")
    print("1. Ver Agenda")
    print("2. Agregar Actividad")
    print("3. Buscar Actividad")
    print("4. Eliminar Actividad")
    print("5. Mostrar Cantidad de Actividades")
    print("6. Ver Primera Actividad")
    print("7. Ver Última Actividad")
    print("8. Ordenar Agenda")
    print("9. Salir")
    print("--------------------------")

    opcion = input("Elige una opción: ")

    if opcion == '1': # Opción: Ver Agenda
        if not agenda:
            print("\nLa agenda está vacía. ¡Añade algunas actividades!")
        else:
            print("\n--- Tu Agenda ---")
            i = 0
            while i < len(agenda):
                print(f"{i + 1}. {agenda[i]}")
                i += 1
            print("-----------------")

    elif opcion == '2': # Opción: Agregar Actividad
        actividad = input("\nIngresa la nueva actividad: ")
        agenda.append(actividad)
        print(f"'{actividad}' ha sido añadida a tu agenda.")

    elif opcion == '3': # Opción: Buscar Actividad
        if not agenda:
            print("\nLa agenda está vacía, no hay nada que buscar.")
        else:
            palabra_clave = input("\nIngresa una palabra clave para buscar: ").lower()

            encontradas = []
            i = 0
            while i < len(agenda):
                if palabra_clave in agenda[i].lower():
                    encontradas.append(agenda[i])
                i += 1

            if encontradas:
                print(f"\n--- Actividades encontradas con '{palabra_clave}' ---")
                i = 0
                while i < len(encontradas):
                    print(f"- {encontradas[i]}")
                    i += 1
                print("-----------------------------------------------\n")
            else:
                print(f"No se encontraron actividades con '{palabra_clave}'.")

    elif opcion == '4': # Opción: Eliminar Actividad
        if not agenda:
            print("\nLa agenda está vacía, no hay nada que eliminar.")
        else:
            print("\n--- Tu Agenda ---")
            i = 0
            while i < len(agenda):
                print(f"{i + 1}. {agenda[i]}")
                i += 1
            print("-----------------")

            try:
                indice_str = input("\nIngresa el número de la actividad que quieres eliminar: ")
                indice = int(indice_str) - 1 # Convertir a índice basado en 0
                if 0 <= indice < len(agenda):
                    actividad_eliminada = agenda.pop(indice)
                    print(f"'{actividad_eliminada}' ha sido eliminada de tu agenda.")
                else:
                    print("Número de actividad inválido.")
            except ValueError:
                print("Por favor, ingresa un número válido.")

    elif opcion == '5': # Opción: Mostrar Cantidad de Actividades
        print(f"\nActualmente tienes {len(agenda)} actividades en tu agenda.")

    elif opcion == '6': # Opción: Ver Primera Actividad
        if agenda:
            print(f"\nLa primera actividad en tu agenda es: '{agenda[0]}'")
        else:
            print("\nLa agenda está vacía.")

    elif opcion == '7': # Opción: Ver Última Actividad
        if agenda:
            print(f"\nLa última actividad en tu agenda es: '{agenda[-1]}'")
        else:
            print("\nLa agenda está vacía.")

    elif opcion == '8': # Opción: Ordenar Agenda
        if not agenda:
            print("\nLa agenda está vacía, no hay nada que ordenar.")
        else:
            agenda.sort() # Método básico de lista para ordenar
            print("\nTu agenda ha sido ordenada alfabéticamente.")

            print("\n--- Tu Agenda (Ordenada) ---")
            i = 0
            while i < len(agenda):
                print(f"{i + 1}. {agenda[i]}")
                i += 1
            print("----------------------------\n")

    elif opcion == '9': # Opción: Salir
        print("\n¡Hasta luego! Gracias por usar la agenda.")
        break # Salir del bucle principal

    else: # Opción no válida
        print("Opción no válida. Por favor, elige un número del 1 al 9.")
```
